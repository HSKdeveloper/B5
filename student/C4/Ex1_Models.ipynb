```python
from typing import Literal
from pydantic import BaseModel, Field

class SentimentResponse(BaseModel):
    """The sentiment of the provided text."""
    sentiment: Literal["positive", "neutral", "negative"] = Field(
        description="The classified sentiment of the input text."
    )
```

```python
# Assuming you are using ChatOpenAI or a similar provider
structured_llm = model_nemotron_nano_creative.with_structured_output(SentimentResponse)
```

```python
# 1. Define the input text
input_text = "Kindness creates lasting joy."

# 2. Invoke the model (this returns an instance of SentimentResponse)
result = structured_llm.invoke(input_text)

# 3. Print the specific sentiment field
print(result.sentiment)
```

```python
from pydantic import BaseModel, Field

class CandidateProfile(BaseModel):
    """Candidate profile details."""
    candidate_name: str = Field(..., description="The name of the candidate")
    skills: list[str] = Field(..., description="A list of skills the candidate possesses")
    experience_in_years: float = Field(..., description="Candidate experience in years")

# Augment the LLM with schema for structured output
structured_llm_candidate = model_nemotron3_nano_precise.with_structured_output(CandidateProfile)

# Example usage (you can modify the input text)
text_input = """
John Doe is a software engineer with expertise in Python, Java, and cloud computing.
He has 5 years of experience in developing scalable web applications and leading small teams.
"""

output = structured_llm_candidate.invoke(text_input)

print("Candidate Name:", output.candidate_name)
print("Skills:", output.skills)
print("Experience:", output.experience_in_years)
```

```python
def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    if b == 0:
        raise ValueError("Cannot divide by zero")
    return a / b
```

```python
system_prompt = """
You are a math assistant. You have access to these tools:

def add(a, b)
def subtract(a, b)
def multiply(a, b)
def divide(a, b)

Return the tool name and arguments required to solve the user's request.
"""
```

```python
from typing import Dict, Any

class ToolNameAndArguments(BaseModel):
    tool_name: str = Field(description="The name of the function to use")
    arguments: Dict[str, Any] = Field(description="The parameters to pass")

tool_calling_llm = model_nemotron3_nano_precise.with_structured_output(ToolNameAndArguments)
```

```python
from langchain.messages import (
    SystemMessage,
    HumanMessage,
    AIMessage
)

response = tool_calling_llm.invoke([
    SystemMessage(system_prompt),
    HumanMessage("what is two plus fiiiive")
])
response
```

```python
if response.tool_name == 'add':
    result = add(response.arguments['a'], response.arguments['b'])

elif response.tool_name == 'subtract':
    result = subtract(response.arguments['a'], response.arguments['b'])

elif response.tool_name == 'multiply':
    result = multiply(response.arguments['a'], response.arguments['b'])

elif response.tool_name == 'divide':
    result = divide(response.arguments['a'], response.arguments['b'])
```

```python
result
```

```python
from langchain.messages import ToolMessage

last = tool_calling_llm.invoke([
    SystemMessage(system_prompt),
    HumanMessage("what is two plus fiiiive"),
    AIMessage(str(response)),
    ToolMessage(
        tool_call_id=123,
        name=response.tool_name,
        arguments=response.arguments
    )
])
last
```
